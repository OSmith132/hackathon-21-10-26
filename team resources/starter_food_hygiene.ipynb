```python
import pandas as pd
import plotly.express as px

DATA = "../02-data-pack/B-food-hygiene/fhrs_subset.csv"
df = pd.read_csv(DATA)
# RatingValue is blank for 'awaiting publication' / not yet rated -> keep as NaN
df["RatingValue"] = pd.to_numeric(df["RatingValue"], errors="coerce")
df
```

## 1. Sanity check (data dive)

Answer these before you build:
- How many rows? How many have a `RatingValue` (vs blank)?
- What is the rating range? (0–5, 5 = best)
- Which `BusinessType` is most common? Which `LocalAuthorityName` do you have?

```python
print("rows:", len(df))
print("rated rows:", df["RatingValue"].notna().sum(),
      "| blank:", df["RatingValue"].isna().sum())
print("rating range:", df["RatingValue"].min(), "->", df["RatingValue"].max())
print("top business types:\n", df["BusinessType"].value_counts().head())
print("local authorities:", df["LocalAuthorityName"].unique())
```

```python
# Example chart: distribution of ratings (5 = best)
rated = df.dropna(subset=["RatingValue"])
px.bar(rated["RatingValue"].value_counts(), title="Number of businesses by hygiene rating")
```
