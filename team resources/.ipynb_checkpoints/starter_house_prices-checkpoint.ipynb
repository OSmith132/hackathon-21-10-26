# Card A — Price Paid Data

**Best archetype:** Predictor / Dashboard. Every property sale in England & Wales (1995–now).

**Your task:** pick ONE — a price **predictor** or a price **dashboard** — for a real citizen (e.g. a first-time buyer choosing a street to live on).

> Contains HM Land Registry data © Crown copyright and database right 2021. Licensed under the Open Government Licence v3.0.

Columns & quirks: `../02-data-pack/A-house-prices/data_dictionary.md`

```python
import pandas as pd
import plotly.express as px

DATA = "../02-data-pack/A-house-prices/price_paid_subset.csv"
COLS = ["transaction_id", "price", "date", "postcode", "type", "new_or_existing",
        "duration", "paon", "saon", "street", "locality", "town", "district",
        "county", "ppd_category", "record_status"]
df = pd.read_csv(DATA, header=None, names=COLS)
df["price"] = pd.to_numeric(df["price"])
df["date"] = pd.to_datetime(df["date"])
df = df[df["record_status"] == "A"]          # A = Added (drop C = Cancelled)
df = df.dropna(subset=["price", "date"])
TYPE = {"T": "Terraced", "D": "Detached", "S": "Semi-detached", "F": "Flat/Maisonette", "O": "Other"}
df["type"] = df["type"].map(TYPE).fillna(df["type"])
df
```

## 1. Sanity check (data dive)

Answer these before you build:
- How many rows? What is the date range?
- Any nulls in `price`, `date`, `type`, `town`?
- What is the median price? How does it differ by `type`?

```python
print("rows:", len(df))
print("date range:", df["date"].min(), "->", df["date"].max())
print("nulls:\n", df[["price", "date", "type", "town"]].isna().sum())
print("median price:", df["price"].median())
df["type"].value_counts()
```

```python
# Example chart: median price over time by property type
df["month"] = df["date"].dt.to_period("M").astype(str)
med = df.groupby(["month", "type"])["price"].median().reset_index()
px.line(med, x="month", y="price", color="type", markers=True)
```

## 2. Your turn — pick ONE and build it

- **Predictor:** estimate `price` from `type` + `town` + `year`. Start with a **baseline** (median by group), then try a linear model.
- **Dashboard:** a 2–3 chart view a citizen would actually use (trend + by type + by area).

Use Feniks to generate code — but **verify every number against the data** before you show it.

```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_absolute_error, mean_squared_error

df["year"] = df["date"].dt.year
print("years in data:", sorted(df["year"].unique()))   # all 2026 in this subset

FEATURES = ["type", "town", "year"]
train, test = train_test_split(df, test_size=0.2, random_state=42)
print(f"train: {len(train):,}   test: {len(test):,}")

def report(y_true, y_pred, name):
    mae = mean_absolute_error(y_true, y_pred)
    rmse = float(np.sqrt(mean_squared_error(y_true, y_pred)))
    print(f"{name:16s}  MAE = \u00a3{mae:,.0f}   RMSE = \u00a3{rmse:,.0f}")
    return mae, rmse

# baseline: median price per (type, town, year) group, computed on TRAIN only
group_dict = dict(train.groupby(FEATURES)["price"].median().items())
town_med = train.groupby("town")["price"].median().to_dict()
type_med = train.groupby("type")["price"].median().to_dict()
overall_med = float(train["price"].median())

def baseline_predict(row):
    key = (row["type"], row["town"], int(row["year"]))
    if key in group_dict:
        return group_dict[key]
    if row["town"] in town_med:
        return town_med[row["town"]]
    if row["type"] in type_med:
        return type_med[row["type"]]
    return overall_med

test["pred_baseline"] = test.apply(baseline_predict, axis=1)
ovr_mae, ovr_rmse = report(test["price"], np.full(len(test), overall_med), "overall median")
base_mae, base_rmse = report(test["price"], test["pred_baseline"], "baseline")
```

```python
from sklearn.linear_model import Ridge
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

# linear model: one-hot the categoricals, keep year numeric; Ridge tames the many town dummies
pre = ColumnTransformer([
    ("cat", OneHotEncoder(handle_unknown="ignore", sparse_output=False), ["type", "town"]),
    ("num", "passthrough", ["year"]),
])
model = Pipeline([("pre", pre), ("ridge", Ridge(alpha=100.0))])
model.fit(train[FEATURES], train["price"])

test["pred_linear"] = model.predict(test[FEATURES])
lin_mae, lin_rmse = report(test["price"], test["pred_linear"], "linear (Ridge)")
```

```python
# citizen-facing estimator: given type, town, year -> predicted price
def estimate_price(type, town, year, method="linear"):
    year = int(year)
    if method == "baseline":
        key = (type, town, year)
        if key in group_dict:
            return group_dict[key]
        if town in town_med:
            return town_med[town]
        if type in type_med:
            return type_med[type]
        return overall_med
    row = pd.DataFrame({"type": [type], "town": [town], "year": [year]})
    return float(model.predict(row)[0])

# example estimates
for t in ["Detached", "Semi-detached", "Terraced", "Flat/Maisonette"]:
    print(f"{t:16s} SWANSEA 2026 -> baseline \u00a3{estimate_price(t, 'SWANSEA', 2026, 'baseline'):,.0f}   linear \u00a3{estimate_price(t, 'SWANSEA', 2026, 'linear'):,.0f}")

# how well does each model do on the held-out test set?
summary = pd.DataFrame({
    "model": ["overall median", "baseline (group med)", "linear (Ridge)"],
    "MAE (\u00a3)": [ovr_mae, base_mae, lin_mae],
    "RMSE (\u00a3)": [ovr_rmse, base_rmse, lin_rmse],
}).set_index("model")
summary
```
