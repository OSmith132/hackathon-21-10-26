## 1. Sanity check (data dive)

- How many rows? What is the month range?
- Which `Crime type` is most common?

```python
print("rows:", len(df))
print("month range:", df["Month"].min(), "->", df["Month"].max())
print("top crime types:\n", df["Crime type"].value_counts().head())
print("unique LSOAs:", df["LSOA name"].nunique())
print("missing coords:", df[["Latitude", "Longitude"]].isna().sum().sum())
```
